## What is a light curve?

Light curves measure the brightness of a star over time, in a graph like this:

```python
plt.figure(figsize=(10,6))      #this line tells the computer we want to make a "figure" or graph
plt.xlabel('time [days]')       #this line makes an x-axis label with the text "time [days]"
plt.ylabel('brightness')        #this line makes an y-axis label with the text "brightness"
plt.xlim(0, 100)                #this line makes the x-axis go from 0 to 100 days
plt.ylim(-100, 1100)            #this line makes the y-axis go from -100 to 1100 units of brightness
plt.tight_layout()
plt.show()
```

### Most stars stay the same brightness for millions of years, so light curves often look very flat over time.

Run the cell below to see what a light curve of a "normal" star might look like!

```python
plt.figure(figsize=(10,6))

# How many times do you want to observe the star? Let's start with 20.
N_observations = 20

# Next we're going to make a list that is N_observations entries long and goes from 0 to 100 days 
times = np.linspace(0, 100, N_observations)

# Now we are going to use a special function called "normal_star"
# This function tells us the brightness of a star if its told the times at which to get its brightness
# To store the brightness values, we have to give them a name! Let's set brightness = normal_star(times).
brightness = normal_star(times)

# Use the function plt.scatter to plot the brightnesses as a scatter plot!
plt.scatter(times, brightness)

plt.xlabel('time [days]')
plt.ylabel('brightness')
plt.ylim(-100, 1100)
plt.tight_layout()
```

You might notice that the data looks pretty messy – the points don't all fall perfectly on the light curve underneath it. When data is imperfect, astronomers call it "noisy." Data may be noisy for a number of reasons. When astronomers measure brightnesses of stars from a photo from a telescope, sometimes light from brighter stars gets in the way of the star we are actually trying to measure. Then, the star appears brighter than it actually is. To stop this from happening, astronomers can measure a smaller area around the star, but sometimes that means astronomers miss some of the light that is actually coming from the supernova. A combination of these and other effects causes the data to be noisy.

```python
fig, ax = plt.subplots(figsize=(10,6))

N_observations = 100
times = np.linspace(0, 100, N_observations)

example_times, example_brightness, example_error = np.loadtxt("./data/sample_supernova_data.txt", unpack=True)

@interact(color = (-0.3, 0.3, 0.01), shape = (-3, 3, 0.1), explosion_time = (15, 45, 1))

def plot(color, shape, explosion_time):
    distance, brightness_model = supernova(times, color = color, shape = shape, explosion_time = explosion_time)
    ax.clear()
    ax.errorbar(example_times, example_brightness, yerr=example_error, ls='None', marker='.', color='black')
    ax.plot(times, brightness_model)
    ax.set_ylim(-100, 1100)
    ax.set_xlim(0, 100)
    ax.set_xlabel('time [days]')
    ax.set_ylabel('brightness')
    
plt.tight_layout()
```

Astronomers can model how energetic a supernova explosion is from our understanding of physics. If a supernova we observe looks fainter than our models, then we know the supernova is further away; if it looks brighter than our models, then we know it is closer to us. ## The Expanding Universe

Measuring distances is really important to map our universe. In addition to knowing the locations of stars and galaxies around our cosmic neighborhood, astronomers want to know how those stars and galaxies are moving around us. By studying the movement of supernovae in the universe, we can gain a better understanding of how the universe is evolving and what the fate of the universe might be!

How would you expect supernovae and the galaxies that they're in to move around the universe? 

### Using a clever but complex technique, astronomers have been able to measure the speed that the galaxies in which supernovae have exploded are moving relative to Earth. Let's now look at the distances and the speeds of our supernovae to learn about how the universe is changing over time.

We will use the distances we measured to SN 2013Rut and SN 2012Car together with the distances measured to 6 other supernovae that have been observed by Hubble. These extra 6 supernova distances are already loaded in your distance table. The speed (or velocity) at which these 8 supernovae are moving is provided in the distance table too. A positive speed means that the supernova is moving away from us, whereas a negative speed means that the supernova is moving towards us. 

How quickly are the supernovae travelling? Do you notice anything that the supernovae have in common in the way they are moving?

```python
fig, ax = plt.subplots(figsize=(10,6))

for snid in distance_table.keys():
    velocity  = distance_table[snid]['velocity']
    distance = distance_table[snid]['distance']
    ax.scatter(velocity, distance, label=snid, marker='*', s=100)

ax.legend()
ax.set_xlabel('velocity [km/sec]')
ax.set_ylabel('distance [Mpc]')
plt.tight_layout()
plt.show()
```

### Wow! Supernovae and the galaxies that they explode in are *not* just moving around the universe randomly; we can see that nearby supernovae are moving away from us slowly, while more distant supernovae are moving away from us quickly! Your graph has just shown that the universe is expanding!
